## 4. Utility functions to reset the face dictionaries and transfer dictionaries by key

```python
def reset_dictionaries(shell):
    faces = Topology.Faces(shell)
    for i, f in enumerate(faces):
        d = Topology.Dictionary(f)
        keys = Dictionary.Keys(d)
        for key in keys:
            if not key == "face_id":
                d = Dictionary.RemoveKey(d, key)
        f = Topology.SetDictionary(f, d)

def transfer_dicts_by_key(topologies, selectors, key):
    dicts = {}
    for t in topologies:
        d = Topology.Dictionary(t)
        value = Dictionary.ValueAtKey(d, key, None)
        if value:
            dicts[str(value)] = t
    
    for s in selectors:
        d = Topology.Dictionary(s)
        value = Dictionary.ValueAtKey(d, key, None)
        if value:
            f = dicts.get(str(value), None)
            if f:
                f = Topology.SetDictionary(f, d)
```

```python
from shapely.geometry import Polygon as ShapelyPolygon
from shapely.geometry import MultiPolygon, GeometryCollection
from shapely.ops import unary_union

def shapely_to_face(polygon):
    """Convert a Shapely Polygon into a Topologic face."""
    if polygon.is_empty or polygon.area <= 1e-8:
        return None
    
    # Attempt to clean minor topology issues
    if not polygon.is_valid:
        polygon = polygon.buffer(0)
        if polygon.is_empty:
            return None
    
    # If cleaning returns MultiPolygon, keep the largest piece
    if polygon.geom_type == "MultiPolygon":
        polygon = max(polygon.geoms, key=lambda g: g.area, default=None)
        if polygon is None or polygon.is_empty:
            return None
    
    outer = Wire.ByVertices(
        [Vertex.ByCoordinates(x, y, 0.0) for x, y in list(polygon.exterior.coords)[:-1]],
        close=True
    )
    holes = [
        Wire.ByVertices(
            [Vertex.ByCoordinates(x, y, 0.0) for x, y in list(ring.coords)[:-1]],
            close=True
        )
        for ring in polygon.interiors
    ]
    return Face.ByWires(outer, holes) if holes else Face.ByWire(outer)

brep_path = r"C:\Users\MOHA9808\Downloads\New folder\GML.26\Asset\resi floor_export_surface.brep"
tri_faces = Topology.Faces(Topology.ByBREPPath(brep_path))

# Build polygons from face vertices and keep only valid, non-degenerate ones
polys = []
for f in tri_faces:
    pts = [(v.X(), v.Y()) for v in Topology.Vertices(f)]
    if len(pts) < 3:
        continue
    p = ShapelyPolygon(pts)
    if not p.is_valid:
        p = p.buffer(0)
    if not p.is_empty and p.area > 1e-8:
        polys.append(p)

if not polys:
    raise ValueError("No valid polygons were extracted from the BREP faces.")

dissolved = unary_union(polys)

# Normalize union output to a list of polygons
if isinstance(dissolved, ShapelyPolygon):
    dissolved_polys = [dissolved]
elif isinstance(dissolved, MultiPolygon):
    dissolved_polys = [p for p in dissolved.geoms if p.area > 1e-8]
elif isinstance(dissolved, GeometryCollection):
    dissolved_polys = [g for g in dissolved.geoms if isinstance(g, ShapelyPolygon) and g.area > 1e-8]
else:
    dissolved_polys = []

parts = [shapely_to_face(p) for p in dissolved_polys]
parts = [(Topology.RemoveCollinearEdges(f) or f) for f in parts if f is not None]

if not parts:
    raise ValueError("Union did not produce any usable polygonal regions.")

gallery = Cluster.ByTopologies(parts) if len(parts) > 1 else parts[0]
```

## 7. Create a grid overlay

```python
b_r = Wire.BoundingRectangle(gallery)
d = Topology.Dictionary(b_r)
xmin = Dictionary.ValueAtKey(d, "xmin")
xmax = Dictionary.ValueAtKey(d, "xmax")
ymin = Dictionary.ValueAtKey(d, "ymin")
ymax = Dictionary.ValueAtKey(d, "ymax")
width = Dictionary.ValueAtKey(d, "width")
length = Dictionary.ValueAtKey(d, "length")

# Halve the original 3-unit cell size to 1.5 units
grid_cell = 1
uCount = int(width / grid_cell) + 2
vCount = int(length / grid_cell) + 2
uRange = [i * grid_cell for i in range(uCount)]
vRange = [i * grid_cell for i in range(vCount)]

grid = Grid.EdgesByDistances(gallery, clip=True, uRange=uRange, vRange=vRange)
```

```python
# ── Community Areas in m² (from notebook variables) ────────────────────────
from collections import defaultdict

community_areas = defaultdict(float)

for f in faces:
    f_dict = Topology.Dictionary(f)
    community_id = Dictionary.ValueAtKey(f_dict, "community", None)
    if community_id is not None:
        area = Face.Area(f)
        community_areas[community_id] += area

# Sort by area descending
sorted_communities = sorted(community_areas.items(), key=lambda x: x[1], reverse=True)

print(f"{'Community':>12}  {'Area (m²)':>12}")
print("-" * 28)
for comm_id, area in sorted_communities:
    print(f"  Community {str(comm_id):>3}  {area:>12.2f} m²")
print("-" * 28)
print(f"{'Total':>12}  {sum(community_areas.values()):>12.2f} m²")
```

### b. Degree centrality

Bin By Dictionary Key
* Use the community partition number (or colour) to separate the faces of the shell into different bins or categories
* Derive the outer boundary of each face group (perimeter) and make a face out of that perimeter

```python
bins = Topology.BinByDictionaryKey(faces, key="community")
bin_dict = bins[0]
keys = list(bin_dict.keys())
face_groups = []
for key in keys:
    bin_faces = bin_dict[key]
    temp_shell = Shell.ByFaces(bin_faces)
    eb = Shell.ExternalBoundary(temp_shell)
    eb = Wire.RemoveCollinearEdges(eb)
    eb = Face.ByWire(eb)
    face_groups.append(eb)

```

```python
for v in g_verts:
    new_v = Vertex.InterpolateValue(v, vertices=new_verts, n=3, key="degree_centrality")
```

Derive the colour of each vertex based on the interpolated value

```python
minValue = min(degree_centralities)
maxValue = max(degree_centralities)
for v in g_verts:
    d = Topology.Dictionary(v)
    d_c = Dictionary.ValueAtKey(d, "degree_centrality")
    color = Color.AnyToHex(Color.ByValueInRange(d_c, minValue=minValue, maxValue=maxValue, colorScale="thermal"))
    d = Dictionary.SetValueAtKey(d, "dc_color", color)
    d = Dictionary.SetValueAtKey(d, "size", 16)
    v = Topology.SetDictionary(v, d)
```
